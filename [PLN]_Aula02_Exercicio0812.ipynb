```python
import spacy
import nltk
import pandas as pd
```

```python
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
```

```python
nlp = spacy.load("pt_core_news_sm")
```

# Pipeline 1

```python
frase_exercicio = "Olá! Gostaria de agendar uma visita para ver o imóvel de 3 quartos amanhã às 14h?"
```

### 1. Tokenização com spaCy

```python
doc_spacy = nlp(frase_exercicio)
tokens_spacy = [token.text for token in doc_spacy]
```

```python
print("Tokens com spaCy:  ", tokens_spacy)
```

### 2. Remoção de pontuação

```python
tokens_filtrados = [token.text for token in doc_spacy if not token.is_punct]
```

### 3. Contagem: original vs. sem pontuação

```python
qtd_original = len(tokens_spacy)
qtd_nao_filtrada = len(tokens_filtrados)
```

```python
print("--- RESULTADOS ---")
print(f"Total de tokens originais: {qtd_original}")
print(f"Total de tokens sem pontuação: {qtd_nao_filtrada}")
```

```python
print("\n--- VISUALIZAÇÃO DOS TOKENS ---")
print(f"Lista Original: {tokens_spacy}")
print(f"Lista Limpa:    {tokens_filtrados}")
```
